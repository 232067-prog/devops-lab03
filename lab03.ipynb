```python
import pandas as pd

df = pd.read_csv('/content/vehicles.csv', engine='python', on_bad_lines='skip')
print(df.shape)
print(df.columns.tolist())
```

```python
# Useless columns drop karo
df = df.drop(columns=['id', 'url', 'region_url', 'VIN', 'image_url', 'description', 'county'])

# Price filter $1000 to $100,000
df = df[(df['price'] >= 1000) & (df['price'] <= 100000)]

# Year filter 2000 to newest
df = df[df['year'] >= 2000]

# Missing values drop
df = df.dropna()

print(df.shape)
```

```python
# Ordinal — manual mapping
condition_map = {'salvage': 1, 'fair': 2, 'good': 3, 'excellent': 4, 'like new': 5, 'new': 6}
df['condition'] = df['condition'].map(condition_map)

# Nominal — one-hot encoding
nominal_cols = ['manufacturer', 'model', 'fuel', 'title_status', 'transmission',
                'drive', 'size', 'type', 'paint_color', 'region', 'state']
df = pd.get_dummies(df, columns=nominal_cols)

print(df.shape)
```

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=['price'])
y = df['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
print(X_train.shape, X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)

print("=== Linear Regression ===")
print(f"R² Score: {r2_score(y_test, y_pred_lr):.4f}")
print(f"MSE: {mean_squared_error(y_test, y_pred_lr):.4f}")
```

```python
from sklearn.tree import DecisionTreeRegressor

dt = DecisionTreeRegressor(random_state=0)
dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)

print("=== Decision Tree ===")
print(f"R² Score: {r2_score(y_test, y_pred_dt):.4f}")
print(f"MSE: {mean_squared_error(y_test, y_pred_dt):.4f}")
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=100, random_state=0)
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)

print("=== Random Forest ===")
print(f"R² Score: {r2_score(y_test, y_pred_rf):.4f}")
print(f"MSE: {mean_squared_error(y_test, y_pred_rf):.4f}")
```

```python
import pandas as pd

results = pd.DataFrame({
    'Model': ['Linear Regression', 'Decision Tree', 'Random Forest'],
    'R² Score': [
        r2_score(y_test, y_pred_lr),
        r2_score(y_test, y_pred_dt),
        r2_score(y_test, y_pred_rf)
    ],
    'MSE': [
        mean_squared_error(y_test, y_pred_lr),
        mean_squared_error(y_test, y_pred_dt),
        mean_squared_error(y_test, y_pred_rf)
    ]
})

print(results)
```
